# Audio Deepfake Detection: Cross-Lingual & Codec Robustness
## Indian Languages Research Walkthrough

This notebook provides a structured research walkthrough investigating:
1. **RQ1:** Cross-lingual generalization to held-out Dravidian languages (Tamil, Telugu, Kannada, Malayalam).
2. **RQ2:** Codec robustness under MP3 and Opus lossy compression.
3. **RQ3:** Accuracy vs efficiency trade-offs across lightweight architectures.

```python
import os
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid", font_scale=1.1)
plt.rcParams["figure.figsize"] = (10, 5)
```

### 1. Load Experimental Results

```python
df_rq1 = pd.read_csv('../results/tables/table1_rq1_cross_lingual.csv')
print("=== RQ1: Cross-Lingual Generalization ===")
display(df_rq1)

df_rq2 = pd.read_csv('../results/tables/table2_rq2_codec_robustness.csv')
print("\n=== RQ2: Codec Robustness ===")
display(df_rq2)

df_eff = pd.read_csv('../results/tables/table3_rq3_efficiency.csv')
print("\n=== RQ3: Computational Efficiency ===")
display(df_eff)
```

### 2. Plot RQ1: Cross-Lingual Generalization Degradation

```python
fig, ax = plt.subplots(figsize=(8, 4))
sns.barplot(data=df_rq1, x='model', y='auroc', hue='split', ax=ax, palette='Blues_r')
ax.set_title('In-Language vs. Held-Out Dravidian AUROC Performance')
ax.set_ylabel('AUROC Score')
ax.set_ylim(0, 1.05)
plt.legend(title='Evaluation Split')
plt.tight_layout()
plt.show()
```

### 3. Plot RQ2: Codec Degradation Profile

```python
fig, ax = plt.subplots(figsize=(8, 4))
x = np.arange(len(df_rq2))
width = 0.35
ax.bar(x - width/2, df_rq2['clean_cnn_auroc'], width, label='Clean-Trained CNN', color='salmon')
ax.bar(x + width/2, df_rq2['augmented_cnn_auroc'], width, label='Codec-Augmented CNN', color='teal')
ax.set_xticks(x)
ax.set_xticklabels(df_rq2['codec_condition'])
ax.set_ylabel('AUROC Score')
ax.set_title('Robustness Across Lossy Transmission Codecs (Dravidian Test Audio)')
ax.legend()
plt.tight_layout()
plt.show()
```
